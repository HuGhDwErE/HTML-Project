```python
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Create a tiny sample dataset of Skyrim items
data = {
    "item_name": [
        "Daedric Boots of Fire Suppression",
        "Ebony Armor of Peerless Destruction",
        "Ring of Eminent Fortify Magicka",
        "Iron Sword of Burning",
        "Glass Shield of Grounding"
    ],
    "description": [
        "Increases Fire Resistance by 50%. Heavy armor boots.",
        "Destruction spells cost 25% less magicka to cast. Heavy torso armor.",
        "Increases total Magicka capacity by 40 points.",
        "Burns the target for 10 points of fire damage.",
        "Increases Shock Resistance by 40%. Light shield."
    ]
}
df = pd.DataFrame(data)

# 2. Load a lightweight, free embedding model (runs locally on your CPU)
# This model converts text into a 384-dimensional vector space.
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Generate vectors for your items by combining their name and description
df['combined_text'] = df['item_name'] + " - " + df['description']
item_embeddings = model.encode(df['combined_text'].tolist())

# 4. Define your search query
user_query = "something to protect me from fire dragons"
query_embedding = model.encode([user_query])

# 5. Run the Exact k-NN Search using Cosine Similarity
similarities = cosine_similarity(query_embedding, item_embeddings)[0]

# 6. Add results to dataframe and sort by highest similarity
df['similarity_score'] = similarities
results = df.sort_values(by='similarity_score', ascending=False)

# Display the top matches
print(f"Results for query: '{user_query}'\n")
print(results[['item_name', 'similarity_score']].to_string(index=False))
```
